# Process raw data to generate ICC data

```python
# This script processes the raw participant data from each corresponding folder to create data files that can be used for the ICC analyses.
```

### Valence and approach-avoidance

```python
# Process the data from valence and approach-avoidance studies (see README file) and save data in a format that can be used for ICC analyses.
data_full  = []
baseline_data_full = []
practice_data_full = []
attention_data_full = []
prolific_ids_full_valence = {}
for index, i in enumerate(os.listdir(raw_file_loc)):
    if i.startswith("V"):
        print(i)
        data, prolific_ids, baseline_data, practice_data, attention_data = process_data(os.path.join(os.path.join(raw_file_loc,i), "labjs_anonymized.csv"),target = "data.valence")
        
        data_full.append(data)
        baseline_data_full.append(baseline_data)
        practice_data_full.append(practice_data)
        attention_data_full.append(attention_data)
        prolific_ids_full_valence.update(prolific_ids)

data_full_valence = pd.concat(data_full)
baseline_data = pd.concat(baseline_data_full)
# Process and save equivalence data
process_equivalence_data(baseline_data=baseline_data,affective_response="valence")
practice_data = pd.concat(practice_data_full) 

# Save CSV file 
icc_data_valence = save_icc_data(data_full_valence, icc_save_loc, ["data.valence", "data.approach"],["valence", "approach"],"data.img_idx",24, True)
```

### Tense arousal and energetic arousal

```python
# Process the data from tense and energetic arousal studies (see README file) and save data in a format that can be used for ICC analyses.
data_full  = []
baseline_data_full = []
practice_data_full = []
attention_data_full = []
prolific_ids_full_arousal = {}
n_participants = 8 #[FIX] I think this is not used again?

for index, i in enumerate(os.listdir(raw_file_loc)):
    if i.startswith("A"):
        print(i)
        data, prolific_ids, baseline_data_arousal, practice_data, attention_data = process_data(os.path.join(os.path.join(raw_file_loc,i), "labjs_anonymized.csv"), target = "data.tense")
        
        data_full.append(data)
        baseline_data_full.append(baseline_data_arousal)
        practice_data_full.append(practice_data)
        attention_data_full.append(attention_data)
        prolific_ids_full_arousal.update(prolific_ids)
    
data_full_arousal = pd.concat(data_full)
# Reverse coding tense arousal ratings!!!
data_full_arousal["data.tense"] = data_full_arousal["data.tense"].map({1:7, 2:6, 3:5, 4:4, 5:3,6:2,7:1})

baseline_data_arousal = pd.concat(baseline_data_full)
# Process and save equivalence data
process_equivalence_data(baseline_data=baseline_data_arousal, affective_response="tense")
practice_data = pd.concat(practice_data_full)

# Save CSV file
icc_data_arousal = save_icc_data(data_full_arousal, icc_save_loc, ["data.tense", "data.energetic"],["tense", "energetic"],"data.img_idx",24, True)
```

## Calculate number of missing ratings for images in IDEAS

```python
import pandas as pd

datasets = {
    'Valence': pd.read_csv(r"..\Data for ICC analysis\icc_valence.csv"),
    'Appraoch': pd.read_csv(r"..\Data for ICC analysis\icc_approach.csv"),
    'Tense': pd.read_csv(r"..\Data for ICC analysis\icc_tense.csv"),
    'Energetic': pd.read_csv(r"..\Data for ICC analysis\icc_energetic.csv")
}

results = []

for name, df in datasets.items():
    # Number of missing ratings per image
    missing_per_image = df.isna().sum(axis=1)

    # Count images with 0, 1, 2, ... missing ratings
    dist = missing_per_image.value_counts().sort_index()

    # Store as a Series with the dataset name
    results.append(dist.rename(name))

# Combine into one table
summary_table = pd.concat(results, axis=1).fillna(0).astype(int)

# Make the index a column if desired
summary_table.index.name = 'Missing Ratings'
summary_table = summary_table.reset_index()

print(summary_table)
```
